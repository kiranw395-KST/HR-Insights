```python
import pandas as pd

salary_df = pd.read_excel("HR_Salary_Stage.xlsx")
demographics_df = pd.read_excel("HR_Demographics_Stage.xlsx")
performance_df = pd.read_excel("HR_Performance_Stage.xlsx")
tenure_df = pd.read_excel("HR_Tenure_Stage.xlsx")
attendance_df = pd.read_excel("HR_Attendance_Stage.xlsx")
HRDATA1_df = pd.read_csv("HR_DATA1.csv")
```

```python
print(salary_df.head())
print(demographics_df.head())
print(performance_df.head())
print(tenure_df.head())
print(attendance_df.head())
print(HRDATA1_df.head())
```

```python
print(demographics_df.head())
```

```python
# Salary distribution
salary_df.groupby('Department')['Avg_Salary'].mean().plot(kind="bar")
```

```python
# Gender ratio
demographics_df.groupby("Sex")["Employee_Count"].sum().plot(kind="pie",autopct="%1.1f%%")
```

```python
# Performance vs Engagement
performance_df.groupby("PerformanceScore")["Avg_Engagement"].mean().plot(kind="bar")
```

```python
# Tenure distribution
tenure_df["Avg_Tenure"].plot(kind="hist", bins=10)
```

```python
# Attendance
attendance_df.plot(x="Department", y=["Avg_Absences","Avg_DaysLate"], kind="bar")
```

```python
salary_df.to_csv("HR_Salary_Clean.csv", index=False)
```

```python
demographics_df.to_csv ("HR_Demographics_Clean.csv", index=False)
performance_df.to_csv ("HR_Performance__Clean.csv", index=False)
tenure_df.to_csv ("HR_Tenure_Stage_Clean.csv", index=False)
attendance_df.to_csv ("HR_Attendance_Clean.csv", index=False)
```
